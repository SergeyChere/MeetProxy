# colab-pilot
Lets your AI agent (Claude Code etc.) run experiments on this GPU.

1. **Runtime → Change runtime type → GPU.**
2. First time only: key icon on the left → add secret **`GH_TOKEN`**: a GitHub fine-grained token for your repo, *Contents: read and write*. Allow notebook access.
3. Put your repo below and **Run all**. Leave the tab open. Stop with ■; shut the runtime down yourself when done.
