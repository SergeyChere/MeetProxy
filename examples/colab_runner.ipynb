# colab-pilot runner
Pick a GPU runtime, add the secret `PILOT_TOKEN` (fine-grained, one private repo, Contents read/write), set the names below, Run all. Stop with the stop button; shut the runtime down yourself when done.

```
MAILBOX = 'you/mailbox'      # private repo with branch 'pilot'
PROJECT = 'you/project'      # repo the jobs work in ('' = none)
PILOT   = 'SergeyChere/colab-pilot'  # this tool
```

```
import os, subprocess
from google.colab import userdata
tok = userdata.get('PILOT_TOKEN')
os.environ['PILOT_TOKEN'] = tok
url = lambda r: f'https://x-access-token:{tok}@github.com/{r}.git'
def sh(c): print(subprocess.run(c, shell=True, capture_output=True, text=True).stdout)
sh('nvidia-smi --query-gpu=name,memory.total --format=csv')
sh(f'rm -rf /content/mailbox /content/pilot; git clone -q -b pilot {url(MAILBOX)} /content/mailbox')
sh(f'git clone -q https://github.com/{PILOT}.git /content/pilot')
if PROJECT: sh(f'[ -d /content/project ] || git clone -q {url(PROJECT)} /content/project')
sh('cd /content/mailbox && git config user.name pilot && git config user.email pilot@localhost')
```
